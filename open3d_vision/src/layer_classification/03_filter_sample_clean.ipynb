# 03 — Extraction et binarisation de la cible

Charge `echantillon_csv` depuis 02, extrait les colonnes cible, binarise (tracer / ne pas tracer).

```python
import numpy as np
import os
```

```python
# Charge l'échantillon nettoyé produit par 02
echantillon_csv = np.load("outputs/echantillon_csv.npy", allow_pickle=True)
print(f"echantillon_csv : {echantillon_csv.shape} (avec colonnes target)")
```

## Extraction et binarisation de la cible

```python
deux_dernieres_colonnes = echantillon_csv[:, -2:].copy()
echantillon_csv = echantillon_csv[:, :-2]

ne_pas_tracer = {"coupe", "terrasses", "cotation", "surface", "texte", "parking", "limite parcellaire", "hachures"}
for i in range(len(deux_dernieres_colonnes)):
    label = str(deux_dernieres_colonnes[i, 1]).strip().lower()
    deux_dernieres_colonnes[i, 0] = 0 if label in ne_pas_tracer else 1

print("Exemple cibles binarisées:", deux_dernieres_colonnes[8:15])

os.makedirs("outputs", exist_ok=True)
np.save("outputs/echantillon_csv.npy", echantillon_csv, allow_pickle=True)
np.save("outputs/deux_dernieres_colonnes.npy", deux_dernieres_colonnes, allow_pickle=True)
print("Sauvegardé: echantillon_csv (features), deux_dernieres_colonnes (cibles binaires)")
```
