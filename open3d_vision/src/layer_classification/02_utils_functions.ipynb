## Chargement, filtrage, échantillonnage et nettoyage

```python
# Charger raw_csv, filtrer, échantillonner
raw_csv = np.load("outputs/raw_csv.npy", allow_pickle=True)
target_csv = raw_csv[raw_csv[:, -1] != "autre"]
del raw_csv  # Libérer la mémoire

np.random.shuffle(target_csv)
echantillon_csv = target_csv[:2000].copy()
del target_csv

# Suppression colonne 1 (trop de valeurs uniques)
echantillon_csv = np.delete(echantillon_csv, 1, 1)

# Unification des types et suppression des colonnes 100 % NaN
echantillon_csv = unifier_types_colonnes(echantillon_csv)
echantillon_csv, _ = colonnes_nan_only(echantillon_csv, supprimer=True)

# Sauvegarde pour 03, 04, 05
import os
os.makedirs("outputs", exist_ok=True)
np.save("outputs/echantillon_csv.npy", echantillon_csv, allow_pickle=True)
print(f"Shape : {echantillon_csv.shape}. Sauvegardé dans outputs/echantillon_csv.npy")
```
