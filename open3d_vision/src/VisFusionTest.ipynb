```python
# Chemins (modifier selon votre setup)
DATA_PATH = ROOT / "data" / "VisFusion-main" / "example_data" / "ScanNet"
CHECKPOINT_PATH = ROOT / "data" / "VisFusion-main"/ "checkpoints" / "pretrained" / "model_000049.ckpt"
SCENE = "scene0785_00"  # None = toutes les scènes test
t = r"C:\Users\mvm\open3d_vision\data\VisFusion-main\checkpoints\pretrained"
# Vérifications
if not DATA_PATH.exists():
    print("ATTENTION: DATA_PATH manquant. Téléchargez l'exemple scene0785_00.")
if not CHECKPOINT_PATH.exists():
    print("ATTENTION: CHECKPOINT_PATH manquant. Téléchargez model_000049.ckpt.")
```

```python
# Exécution VisFusion
from visfusion_runner import run_visfusion

result = run_visfusion(
    scene=SCENE,
    data_path=DATA_PATH,
    checkpoint_path=CHECKPOINT_PATH if CHECKPOINT_PATH.exists() else None,
    single_layer_mesh=False,  # True pour format TransformerFusion
)

print("Code de sortie:", result.returncode)
# Les meshes sont sauvegardés dans data/VisFusion-main/results/
```

```python
# Visualisation du mesh avec Open3D (optionnel)
results_dir = ROOT / "data" / "VisFusion-main" / "results"
mesh_files = list(results_dir.glob("**/*.ply")) if results_dir.exists() else []

if mesh_files:
    import open3d as o3d
    mesh = o3d.io.read_triangle_mesh(str(mesh_files[-1]))
    print(f"Mesh chargé: {mesh_files[-1].name}, vertices: {len(mesh.vertices)}")
    o3d.visualization.draw_geometries([mesh], window_name="VisFusion - Reconstruction")
else:
    print("Aucun mesh trouvé. Exécutez la cellule précédente après téléchargement des données.")
```
